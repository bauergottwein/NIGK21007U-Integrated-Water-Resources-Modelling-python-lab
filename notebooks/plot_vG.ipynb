This notebook shows you how you can plot the van Genuchten retention curve:

```python
import numpy as np
import matplotlib.pyplot as plt

def vG(psi,thetar,thetas,alpha,n):
    S = (1+(alpha*psi)**n)**((1/n)-1)
    theta = S*(thetas-thetar) + thetar
    return theta

psilog = np.linspace(-1,4,50)
psi = 10**psilog

theta = vG(psi,0.05,0.4,0.4,1.1)
plt.plot(theta,psilog,'g')
theta = vG(psi,0.05,0.4,0.4,1.5)
plt.plot(theta,psilog,'b')
theta = vG(psi,0.05,0.4,0.4,4)
plt.plot(theta,psilog,'r')
theta = vG(psi,0.05,0.4,0.8,4)
plt.plot(theta,psilog,'k')
plt.xlabel('Soil moisture content')
plt.ylabel('log of Matric potential (m)')
plt.show()
```
